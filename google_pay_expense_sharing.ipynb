**prettytable** - representing output in a table structure

```python
import numpy as np
import prettytable
```

```python
friends=['Ashley','Zia','Deepu','Rohit','Sahil']
n=len(friends)
print("Total Friends:",n)
```

```python
expense_matrix= np.zeros((len(friends),len(friends)))
expense_matrix
```

*   row - payer
*   column - benefiaries

```python
def add_expense(payer,beneficiaries,amount):
  payer_index=friends.index(payer)
  share_per_person=amount/len(beneficiaries)
  for b in beneficiaries:
    ben_idx=friends.index(b)
    expense_matrix[payer_index][ben_idx]+=share_per_person
```

```python
def calculate_settlements():
  total_paid=np.sum(expense_matrix,axis=1) #amount each person paid (columns)
  total_owes=np.sum(expense_matrix,axis=0) #amount each person owes (rows)
  net_balance=total_paid-total_owes
  return net_balance
```

```python
def display_settlements():
  settlements=calculate_settlements()

  table=prettytable.PrettyTable()
  table.field_names=["Friend","Settlement"]

  for i, friend in enumerate(friends):
    if(settlements[i]>0):
        table.add_row([friend,f"should receive Rs.{settlements[i]:.2f}"])
    elif(settlements[i]<0):
        table.add_row([friend,f"owes Rs.{-settlements[i]:.2f}"])
    else:
      table.add_row([friend,"is settled"])

  print("Final Settlements:")
  print(table)
```

```python
def suggest_payments():
  settlements=calculate_settlements()

  creditors=[(friends[i],amt) for i,amt in enumerate(settlements) if amt>0]
  debtors=[(friends[i],-amt) for i,amt in enumerate(settlements) if amt<0]

  transactions=[]

  while debtors and creditors:
    d, debt_amt=debtors.pop()
    c, credit_amt=creditors.pop()

    payment=min(debt_amt,credit_amt)
    transactions.append((d,c,payment))

    debt_amt-=payment
    credit_amt-=payment

    if debt_amt>0:
      debtors.insert(0,(d,debt_amt))
    if credit_amt>0:
      creditors.insert(0,(c,credit_amt))

  print("Suggested Transactions:")
  if transactions:
    for d,c,amount in transactions:
      print(f"{d} should pay {amount:.2f} to {c}")
  else:
    print("No transactions needed. Everyone is settled.")
```

```python
add_expense('Sahil',['Ashley','Zia'],2000)
display_settlements()
suggest_payments()
```

```python
add_expense('Ashley',['Deepu','Rohit','Zia'],1500)
display_settlements()
suggest_payments()
```

```python
add_expense('Deepu',['Sahil','Zia'],1500)
display_settlements()
suggest_payments()
```
